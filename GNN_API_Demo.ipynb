# GNN API Demo Notebook

This notebook demonstrates how to interact with your FastAPI + GNN backend for malware classification. It shows how to send a sample request to the API, process the response, and display the result.

```python
# Import Required Libraries
import requests
import json
```

## Define Sample Input for GNN Prediction
Below, we define a sample graph input (features, edges, node count) to send to the API.

```python
# Define a sample graph input for the GNN API
sample_payload = {
    "features": [[0.1]*1433, [0.2]*1433, [0.3]*1433],  # 3 nodes, 1433 features each (Cora dataset)
    "edges": [[0, 1], [1, 2]],  # Simple edge list
    "node_count": 3
}
print(json.dumps(sample_payload, indent=2)[:500] + '...')  # Print a preview
```

## Send Request to GNN API and Get Prediction
This cell sends the sample input to the FastAPI backend and prints the prediction result.

```python
# Send the request to the FastAPI GNN backend
api_url = "http://127.0.0.1:8000/api/gnn/predict"
try:
    response = requests.post(api_url, json=sample_payload)
    result = response.json()
    print("Prediction result:", json.dumps(result, indent=2))
except Exception as e:
    print("Error contacting API:", e)
```
